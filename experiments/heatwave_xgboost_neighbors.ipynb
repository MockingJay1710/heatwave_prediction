# Expérience 2 — XGBoost et quatre voisins

Est-ce que les conditions au nord, au sud, à l’est et à l’ouest améliorent le modèle temporel de l’expérience 1 ?


On prédit **Tmax de demain** à partir de la météo d'aujourd'hui.
On suppose donc être en fin de journée, une fois toutes les observations du jour connues.
Les journées sont définies en UTC. Les données ERA5 sont rétrospectives : ce n'est pas
encore une simulation de prévision météo en temps réel.

Tout le code de l'expérience est dans ce notebook. Si le CSV manque, lancer depuis
la racine du projet : `nix develop --command python scripts/download_four_neighbors.py`.


```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error
from IPython.display import display
```

## 1. Charger les données

Le CSV contient Nantes et quatre voisins. On commence avec Nantes seulement.


```python
DATA_PATH = Path("../data/weather_five_points_daily.csv")
toutes_les_donnees = pd.read_csv(DATA_PATH, parse_dates=["date"])

display(toutes_les_donnees[["point", "lat", "lon"]].drop_duplicates())

nantes = toutes_les_donnees[toutes_les_donnees["point"] == "nantes"].copy()
nantes = nantes.sort_values("date").reset_index(drop=True)

assert not toutes_les_donnees.duplicated(["point", "date"]).any()
assert not toutes_les_donnees.isna().any().any()

print(f"{len(nantes)} journées, du {nantes['date'].min().date()} au {nantes['date'].max().date()}")
display(nantes.head())
```

## 2. Construire la cible : la Tmax du lendemain

Pour chaque date J, on cherche la température à la date **J + 1 jour**.
Cette correspondance sur la date évite de confondre « ligne suivante » et « lendemain »
si un jour manque. La dernière journée est retirée car son lendemain n'est pas connu.


```python
temperature_par_date = nantes.set_index("date")["temperature_max"]

nantes["date_cible"] = nantes["date"] + pd.Timedelta(days=1)
nantes["tmax_demain"] = nantes["date_cible"].map(temperature_par_date)
nantes = nantes.dropna(subset=["tmax_demain"]).copy()

display(nantes[["date", "temperature_max", "date_cible", "tmax_demain"]].head())
```

## 3. Ajouter le jour de l'année et la saison

On utilise le calendrier de **demain**, dont on connaît déjà la date.
Le jour de l'année est représenté par un sinus et un cosinus : fin décembre reste
proche de début janvier. Le calcul tient compte des années à 365 ou 366 jours.

La saison est représentée par quatre colonnes 0/1. On n'ajoute ni mois ni année
comme variables supplémentaires.


```python
dates = nantes["date_cible"]
nantes["jour_annee"] = dates.dt.dayofyear

jours_dans_annee = np.where(dates.dt.is_leap_year, 366, 365)
angle = 2 * np.pi * (nantes["jour_annee"] - 1) / jours_dans_annee
nantes["jour_sin"] = np.sin(angle)
nantes["jour_cos"] = np.cos(angle)

saisons_par_mois = {
    12: "hiver", 1: "hiver", 2: "hiver",
    3: "printemps", 4: "printemps", 5: "printemps",
    6: "ete", 7: "ete", 8: "ete",
    9: "automne", 10: "automne", 11: "automne",
}
nantes["saison"] = dates.dt.month.map(saisons_par_mois)
saisons = pd.get_dummies(nantes["saison"], prefix="saison")
nantes = pd.concat([nantes, saisons], axis=1)

display(nantes[["date_cible", "jour_annee", "jour_sin", "jour_cos", "saison"]].head())
```

## 4. Ajouter les quatre voisins

On garde exactement les mêmes observations de Nantes. Pour chaque direction, on ajoute
les moyennes du jour J : température, pression et composantes du vent.
La jointure utilise la **date** ; aucune valeur du lendemain n'est ajoutée.


```python
directions = ["nord", "sud", "est", "ouest"]

for direction in directions:
    voisin = toutes_les_donnees[toutes_les_donnees["point"] == direction].copy()
    voisin = voisin[["date", "temperature_mean", "pressure_msl_mean", "u_mean", "v_mean"]]
    voisin = voisin.rename(columns={
        "temperature_mean": f"temperature_{direction}",
        "pressure_msl_mean": f"pression_{direction}",
        "u_mean": f"u_{direction}",
        "v_mean": f"v_{direction}",
    })
    nantes = nantes.merge(voisin, on="date", how="left", validate="one_to_one")

assert not nantes.isna().any().any()
display(nantes[["date", "temperature_mean", "temperature_nord", "temperature_sud", "temperature_est", "temperature_ouest"]].head())
```

## 5. Calculer les contrastes de température

Pour chaque direction : **température du voisin − température de Nantes**.
Un delta négatif indique de l'air plus froid dans cette direction.
On utilise partout des températures moyennes du même jour.


```python
for direction in directions:
    nantes[f"delta_temperature_{direction}"] = (
        nantes[f"temperature_{direction}"] - nantes["temperature_mean"]
    )

display(nantes[["date", "delta_temperature_nord", "delta_temperature_sud", "delta_temperature_est", "delta_temperature_ouest"]].head())
```

## 6. Comprendre le vecteur vent

La direction météo indique **d'où vient le vent**. Avec une vitesse V et une direction θ en radians :

\[
u=-V\sin(\theta),\qquad v=-V\cos(\theta)
\]

- u positif : vent qui se déplace vers l'est.
- v positif : vent qui se déplace vers le nord.

Le script de collecte calcule u et v **pour chaque heure**, puis leur moyenne par jour.
C'est plus cohérent que multiplier la vitesse moyenne par une direction moyenne.
Les colonnes `u_mean`, `v_mean` concernent Nantes ; `u_nord`, `v_nord`, etc. concernent les voisins.


```python
display(nantes[["date", "u_mean", "v_mean", "u_nord", "v_nord", "u_ouest", "v_ouest"]].head())
```

## 7. Approximer le gradient de pression

On calcule deux pentes : **(pression est − pression ouest) / distance est-ouest**,
et **(pression nord − pression sud) / distance nord-sud**.
Les distances sont en kilomètres ; les gradients seront donc en **hPa/km**.

On utilise les coordonnées réelles des cellules. À cette échelle, on approxime un degré
de latitude par 111,32 km ; pour la longitude, on multiplie aussi par le cosinus de la latitude.


```python
positions = toutes_les_donnees[["point", "lat", "lon"]].drop_duplicates().set_index("point")
latitude_nantes = positions.loc["nantes", "lat"]

distance_est_ouest_km = (
    111.32 * np.cos(np.deg2rad(latitude_nantes))
    * (positions.loc["est", "lon"] - positions.loc["ouest", "lon"])
)
distance_nord_sud_km = (
    111.32 * (positions.loc["nord", "lat"] - positions.loc["sud", "lat"])
)

nantes["gradient_pression_est_ouest"] = (
    nantes["pression_est"] - nantes["pression_ouest"]
) / distance_est_ouest_km
nantes["gradient_pression_nord_sud"] = (
    nantes["pression_nord"] - nantes["pression_sud"]
) / distance_nord_sud_km

print(f"Distance est-ouest : {distance_est_ouest_km:.1f} km")
print(f"Distance nord-sud : {distance_nord_sud_km:.1f} km")
display(nantes[["date", "gradient_pression_est_ouest", "gradient_pression_nord_sud"]].head())
```

## 8. Choisir les variables

Le premier modèle reprend l’expérience 1. Le second ajoute les voisins, leurs contrastes de température, les vents et les deux gradients de pression.


```python
colonnes_meteo = [
    "temperature_max", "temperature_min", "temperature_mean",
    "relative_humidity_mean", "dew_point_mean", "pressure_msl_mean",
    "cloud_cover_mean", "wind_speed_mean", "wind_speed_max",
    "precipitation_sum", "soil_moisture_mean", "shortwave_radiation_sum_MJ_m2",
]
colonnes_temps = ["jour_sin", "jour_cos"] + list(saisons.columns)
colonnes_temporelles = colonnes_meteo + colonnes_temps
```

```python
colonnes_voisinage = [
    "u_mean", "v_mean",
    "gradient_pression_est_ouest", "gradient_pression_nord_sud",
]
for direction in directions:
    colonnes_voisinage += [
        f"temperature_{direction}", f"pression_{direction}",
        f"u_{direction}", f"v_{direction}",
        f"delta_temperature_{direction}",
    ]

colonnes_completes = colonnes_temporelles + colonnes_voisinage
print(f"Modèle temporel : {len(colonnes_temporelles)} variables")
print(f"Avec voisinage : {len(colonnes_completes)} variables")
```

## 9. Reprendre la même séparation chronologique

Les journées et la date de début du test sont identiques à celles de l’expérience 1.


```python
# Les 80 % premières dates pour apprendre, les 20 % dernières pour évaluer.
position_coupure = int(len(nantes) * 0.8)
date_test = nantes.iloc[position_coupure]["date_cible"]

train = nantes[nantes["date_cible"] < date_test].copy()
test = nantes[nantes["date_cible"] >= date_test].copy()

y_train = train["tmax_demain"]
y_test = test["tmax_demain"]
persistance = test["temperature_max"]

print(f"Entraînement : {len(train)} jours, jusqu'au {train['date_cible'].max().date()}")
print(f"Test : {len(test)} jours, à partir du {test['date_cible'].min().date()}")
```

## 10. Entraîner les deux XGBoost

On conserve les mêmes paramètres que dans l’expérience 1.


```python
modele_temporel = XGBRegressor(
    n_estimators=300,
    max_depth=3,
    learning_rate=0.05,
    objective="reg:squarederror",
    tree_method="hist",
    random_state=42,
    n_jobs=2,
)
modele_temporel.fit(train[colonnes_temporelles], y_train)
prediction_temporelle = modele_temporel.predict(test[colonnes_temporelles])
```

```python
modele_voisins = XGBRegressor(
    n_estimators=300,
    max_depth=3,
    learning_rate=0.05,
    objective="reg:squarederror",
    tree_method="hist",
    random_state=42,
    n_jobs=2,
)
modele_voisins.fit(train[colonnes_completes], y_train)
prediction_voisins = modele_voisins.predict(test[colonnes_completes])
```

## 11. Comparer les erreurs


```python
scores = pd.DataFrame({
    "modele": ["Persistance", "XGBoost + jour et saison", "XGBoost + calendrier et voisins"],
    "MAE_C": [
        mean_absolute_error(y_test, persistance),
        mean_absolute_error(y_test, prediction_temporelle),
        mean_absolute_error(y_test, prediction_voisins),
    ],
})
display(scores.round(3))
```

```python
plt.figure(figsize=(15, 5))
plt.plot(test["date_cible"], y_test, label="Tmax réelle", color="black")
plt.plot(test["date_cible"], persistance, label="Persistance", alpha=0.4)
plt.plot(test["date_cible"], prediction_temporelle, label="Jour et saison", alpha=0.8)
plt.plot(test["date_cible"], prediction_voisins, label="Avec les quatre voisins", alpha=0.8)
plt.xlabel("Date à prédire")
plt.ylabel("Tmax (°C)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
plt.close()
```

## À retenir

Une MAE plus basse signifie une meilleure prévision moyenne. Les modèles utilisent
les mêmes dates de test et les mêmes paramètres : on change seulement les variables d'entrée.
La persistance signifie « demain aura la même Tmax qu'aujourd'hui ».

On garde ces paramètres fixés, sans les modifier à partir des résultats du test.
Ces dates ayant déjà été regardées lors d'essais précédents, les résultats restent exploratoires.
